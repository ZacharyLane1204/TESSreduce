```python
import numpy as np
from scipy.ndimage import gaussian_filter1d

def adaptive_gaussian_smooth(time, bkg, sigma_min_hrs=0.5, sigma_max_hrs=6,
                              n_levels=15, gap_threshold_hrs=1.0,
                              raw_grad_threshold=0.3, spike_nsigma=5):
    '''
    Adaptive Gaussian smoothing in physical time units, respecting data gaps.

    Points are left as raw data if they have a high local gradient OR are
    significantly elevated above the local baseline (catches full spike decay).

    Parameters
    ----------
    time               : 1D array of times (days)
    bkg                : 1D array of background values
    sigma_min_hrs      : minimum smoothing width in hours (near rapid changes)
    sigma_max_hrs      : maximum smoothing width in hours (steady regions)
    n_levels           : number of pre-computed sigma levels
    gap_threshold_hrs  : gaps larger than this split data into independent segments
    raw_grad_threshold : gradient fraction [0-1] above which points are left raw
    spike_nsigma       : points more than this many sigma above local baseline
                         are also left raw
    '''
    dt_hrs    = np.median(np.diff(time)) * 24
    sigma_min = sigma_min_hrs / dt_hrs
    sigma_max = sigma_max_hrs / dt_hrs

    gaps       = np.diff(time) * 24
    seg_breaks = np.where(gaps > gap_threshold_hrs)[0] + 1
    seg_starts = np.concatenate([[0], seg_breaks])
    seg_ends   = np.concatenate([seg_breaks, [len(bkg)]])

    result = bkg.copy()

    for s, e in zip(seg_starts, seg_ends):
        seg    = bkg[s:e]
        sigmas = np.geomspace(sigma_min, sigma_max, n_levels)

        # Robust baseline and noise estimate from the lower 60th percentile
        lo      = seg <= np.percentile(seg, 60)
        baseline = np.median(seg[lo])
        noise    = np.median(np.abs(seg[lo] - baseline))
        noise    = max(noise, 1e-6)

        # Gradient-based raw mask
        grad      = np.abs(np.gradient(seg))
        p_hi      = np.percentile(grad, 99)
        grad_norm = np.clip(grad / (p_hi + 1e-10), 0, 1)

        # Combined raw mask: high gradient OR elevated above baseline
        raw_mask    = (grad_norm >= raw_grad_threshold) | (seg > baseline + spike_nsigma * noise)
        smooth_mask = ~raw_mask

        # Sigma index: high gradient -> sigma_min, flat -> sigma_max
        sigma_idx = (1.0 - grad_norm ** 2) * (n_levels - 1)
        sigma_idx = gaussian_filter1d(sigma_idx, sigma=max(1, 1.0 / dt_hrs))
        sigma_idx = np.clip(sigma_idx, 0, n_levels - 1)

        # Pre-compute smoothed versions at each sigma level
        smoothed = np.array([gaussian_filter1d(seg, sigma=sig, mode='nearest')
                             for sig in sigmas])

        # Interpolate between adjacent levels
        idx_lo  = np.clip(sigma_idx.astype(int), 0, n_levels - 2)
        idx_hi  = idx_lo + 1
        frac    = sigma_idx - idx_lo
        val_lo  = np.take_along_axis(smoothed, idx_lo[np.newaxis], axis=0)[0]
        val_hi  = np.take_along_axis(smoothed, idx_hi[np.newaxis], axis=0)[0]
        blended = (1.0 - frac) * val_lo + frac * val_hi

        seg_result = seg.copy()
        seg_result[smooth_mask] = blended[smooth_mask]
        result[s:e] = seg_result

    return result
```

```python
"""
TESSSourceFinder
================
Multi-scale Mexican-hat SEP source detection for TESS reference images.

Best-fit parameters derived from Gaia recall optimisation on sector 38,
cam 4, ccd 1 (20.5 arcsec/px, 256x256 cut):

    83% Gaia recall (G < 18.5), median position error 0.95 px
    100% recall for G < 14, sub-0.2 px position accuracy
"""

import numpy as np
import pandas as pd
import sep
import warnings
warnings.filterwarnings("ignore")


class TESSSourceFinder:
    """
    Detect point sources in a TESS reference image using multi-scale
    Mexican-hat convolution kernels and SEP (Source Extractor Python).

    Parameters
    ----------
    thresh : float
        Detection threshold in units of background RMS. Default 0.30.
    bw : int
        Background mesh cell size in pixels. Default 24.
    fwhm_min, fwhm_max : float
        Range of kernel FWHM values (pixels) to sweep. Default 0.7–2.0.
    n_scales : int
        Number of kernel scales between fwhm_min and fwhm_max. Default 10.
    deblend_nthresh : int
        SEP deblending threshold levels. Default 128.
    deblend_cont : float
        SEP minimum contrast ratio for deblending. Default 5e-5.
    dedup_radius : float
        Pixels — detections from different scales within this distance are
        deduplicated, keeping the one with the highest flux. Default 1.0.
    ref : ndarray (H, W), optional
        Reference image (tessreduce .ref).  When provided, a per-pixel
        noise map is built as noise[y,x] = sqrt(bkg_rms² + (k*ref[y,x])²)
        and used as the SEP error map.
    residual_k : float, optional
        Fractional systematic residual coefficient.  Typical TESS value
        ~0.0078.  If None, estimated automatically from the ref image.
    """

    def __init__(
        self,
        thresh: float = 0.30,
        bw: int = 24,
        fwhm_min: float = 0.7,
        fwhm_max: float = 2.0,
        n_scales: int = 10,
        deblend_nthresh: int = 128,
        deblend_cont: float = 5e-5,
        dedup_radius: float = 1.0,
        ref: np.ndarray = None,
        residual_k: float = None,
    ):
        self.thresh = thresh
        self.bw = bw
        self.fwhm_min = fwhm_min
        self.fwhm_max = fwhm_max
        self.n_scales = n_scales
        self.deblend_nthresh = deblend_nthresh
        self.deblend_cont = deblend_cont
        self.dedup_radius = dedup_radius
        self.ref = np.asarray(ref, dtype=np.float64) if ref is not None else None
        self.residual_k = residual_k

        self._fwhms = np.linspace(fwhm_min, fwhm_max, n_scales)
        self._kernels = [self._mexhat(f) for f in self._fwhms]

    # ── public API ────────────────────────────────────────────────────────────

    def find(self, image: np.ndarray) -> pd.DataFrame:
        """
        Detect sources in a 2-D image or a 3-D timeseries.

        Parameters
        ----------
        image : ndarray (H, W) or (T, H, W)
            For a single image, returns the standard detection DataFrame.
            For a timeseries, detection is parallelised across frames with
            joblib and a 'frame' column (0-indexed) is added.

        Returns
        -------
        sources : DataFrame
            Columns: x, y, flux, peak, a, b, theta, flag, detection_fwhm,
            snr, ellipticity, neg_extent [, snr_model if ref provided]
            [, frame if timeseries input]
        """
        image = np.asarray(image)
        if image.ndim == 2:
            return self._find_single(image)
        elif image.ndim == 3:
            return self._find_cube(image)
        else:
            raise ValueError(f"image must be 2D or 3D, got shape {image.shape}")

    def filter(self, sources: pd.DataFrame,
               snr_min: float = 2.0,
               snr_model_min: float = 2.0,
               ellipticity_max: float = 0.6,
               neg_extent_max: float = 6,
               peak_min: float = None,
               flux_min: float = None) -> pd.DataFrame:
        """
        Apply quality filters to remove contamination from differenced-image
        detections while preserving real astrophysical sources.

        Contamination taxonomy (from analysis of TESS sector 38 diff cube):
        ┌──────────────────────────────────┬──────────────────────────────────────┐
        │ Type                             │ Signature                            │
        ├──────────────────────────────────┼──────────────────────────────────────┤
        │ Ringing artefact (60%)           │ snr < 0, neg_extent > 5              │
        │   Mexican-hat fires on positive  │ Aperture captures surrounding        │
        │   peak of a bright star's halo   │ negative bowl from PSF halo          │
        ├──────────────────────────────────┼──────────────────────────────────────┤
        │ Faint halo artefact (24%)        │ snr < 0, neg_extent 2–5              │
        │   Weaker version of above        │                                      │
        ├──────────────────────────────────┼──────────────────────────────────────┤
        │ Cosmic ray / scattered light     │ snr_model >> 10 on dim sky,          │
        │   cluster artefact               │ snr < 0, multiple nearby detections  │
        └──────────────────────────────────┴──────────────────────────────────────┘

        Parameters
        ----------
        sources : DataFrame
            Output of find().  If ref was not provided, snr_model will be
            absent and that criterion is skipped.
        snr_min : float
            Minimum aperture SNR.  Default 2.0.
        snr_model_min : float
            Minimum peak / ref-scaled noise.  Skipped if snr_model column is
            absent (i.e. no ref image was provided).  Default 2.0.
        ellipticity_max : float
            Maximum ellipticity (1 − b/a).  Default 0.6.
        neg_extent_max : float or None
            Maximum neg_extent (sigma-normalised negative bowl depth).
            None = no cut.
        peak_min : float or None
            Minimum peak pixel value above background.  None = no cut.
        flux_min : float or None
            Minimum isophotal flux above background.  None = no cut.

        Returns
        -------
        filtered : DataFrame, reset index.
        """
        mask = pd.Series(True, index=sources.index)

        mask &= sources["snr"] >= snr_min
        mask &= sources["ellipticity"].fillna(1.0) < ellipticity_max

        if "snr_model" in sources.columns:
            mask &= sources["snr_model"] >= snr_model_min

        if neg_extent_max is not None and "neg_extent" in sources.columns:
            mask &= sources["neg_extent"].fillna(0.0) <= neg_extent_max

        if peak_min is not None:
            mask &= sources["peak"] >= peak_min

        if flux_min is not None:
            mask &= sources["flux"] >= flux_min

        mask &= (sources['flag'] == 0) | (sources['flag'] == 1)
        
        return sources[mask].reset_index(drop=True)

    def group(self, detections: pd.DataFrame,
              spatial_eps: float = 1.5,
              max_speed: float = 0.3,
              event_gap: int = 10) -> pd.DataFrame:
        """
        Cluster detections into stationary sources using DBSCAN on (x, y),
        then purge moving objects that were absorbed into spatial clusters.

        Problem with pure spatial clustering: an asteroid crossing near a star
        will have its detections folded into that star's cluster (they share the
        same sky region for a few frames), so the asteroid never appears as an
        unassigned detection and is invisible to find_asteroids().

        Fix — two-pass consistency check:
          Pass 1: spatial DBSCAN groups detections that are co-located in (x,y)
                  regardless of time.  This correctly groups stationary stars.
          Pass 2: for every multi-frame cluster, compute the per-frame SNR-
                  weighted centroid and fit a linear velocity.  If the inferred
                  speed exceeds max_speed px/frame the cluster contains a moving
                  object.  Detections whose residual from the star's SNR-weighted
                  mean position exceeds spatial_eps are re-labelled as unassigned
                  (source_id = -1) so find_asteroids() can pick them up.

        Parameters
        ----------
        detections : DataFrame
            Output of find() on a 3-D timeseries.
        spatial_eps : float
            DBSCAN neighbourhood radius in pixels.  Default 1.5.
        min_frames : int
            Minimum number of distinct frames a cluster must span to be
            labelled as a stationary source.  Default 2.
        max_speed : float
            Maximum allowed drift of a cluster's per-frame centroid in
            px/frame.  Clusters with fitted velocity exceeding this have their
            off-position detections split out as unassigned.  Default 0.3.
        event_gap : int
            Frame gap that separates distinct events for the same stationary
            source.  Within a source group, any pair of consecutive detections
            separated by more than event_gap frames starts a new event.
            Default 10 frames.

        Returns
        -------
        detections : DataFrame with added columns:
            source_id — int cluster label (>= 0 stationary, -1 unassigned)
            event_id  — per-source event index (0, 1, 2, …) ordered in time;
                        NaN for unassigned detections (source_id == -1)
            x_w, y_w  — SNR-weighted centroid of the stationary source this
                        detection belongs to; copies raw x, y for unassigned.
        """
        if "frame" not in detections.columns:
            raise ValueError("detections must have a 'frame' column")

        from sklearn.cluster import DBSCAN

        xy = detections[["x", "y"]].values
        labels = DBSCAN(eps=spatial_eps, min_samples=1).fit_predict(xy)

        detections = detections.copy()
        detections["source_id"] = labels

        # ── Motion consistency check ──────────────────────────────────────
        # For each stationary candidate, compute per-frame SNR-weighted
        # centroid and fit a velocity.  Detections that are offset from the
        # time-averaged star position by more than spatial_eps are split out.
        w_col = "snr_model" if "snr_model" in detections.columns else "snr"
        weights = detections[w_col].clip(lower=1e-3).values

        moving_rows = []  # integer positional indices to demote

        for sid, idx in detections.groupby("source_id").groups.items():
            if sid == -1:
                continue
            sub = detections.loc[idx]
            w = weights[idx]

            # SNR-weighted centroid (the star's expected fixed position)
            w_sum = w.sum()
            cx = (sub["x"].values * w).sum() / w_sum
            cy = (sub["y"].values * w).sum() / w_sum

            # Per-frame centroid to measure drift
            frames = sub["frame"].values
            if len(np.unique(frames)) < 2:
                continue

            # Fit velocity to per-frame SNR-weighted centroids
            frame_cx = {}
            frame_cy = {}
            for fi in np.unique(frames):
                mask = frames == fi
                wi = w[mask]
                ws = wi.sum()
                frame_cx[fi] = (sub["x"].values[mask] * wi).sum() / ws
                frame_cy[fi] = (sub["y"].values[mask] * wi).sum() / ws

            ft = np.array(sorted(frame_cx))
            fx = np.array([frame_cx[f] for f in ft])
            fy = np.array([frame_cy[f] for f in ft])

            if len(ft) >= 2:
                vx = np.polyfit(ft, fx, 1)[0]
                vy = np.polyfit(ft, fy, 1)[0]
                speed = np.hypot(vx, vy)
            else:
                speed = 0.0

            if speed > max_speed:
                # Cluster is drifting — identify which detections are outliers
                # from the star's SNR-weighted fixed position and demote them.
                dist_from_star = np.hypot(sub["x"].values - cx,
                                          sub["y"].values - cy)
                outlier_mask = dist_from_star > spatial_eps
                pos_idx = detections.index.get_indexer(idx)
                moving_rows.extend(pos_idx[outlier_mask].tolist())

        if moving_rows:
            iloc_arr = np.array(moving_rows)
            detections.iloc[iloc_arr, detections.columns.get_loc("source_id")] = -1

        # ── Event labelling ───────────────────────────────────────────────
        # Within each stationary source, split detections into events by
        # looking for gaps in the frame sequence larger than event_gap.
        # A new event starts whenever the frame jump exceeds the threshold.
        event_id_col = np.full(len(detections), np.nan)

        for sid, idx in detections.groupby("source_id").groups.items():
            if sid == -1:
                continue
            sub_frames = detections.loc[idx, "frame"].values
            order = np.argsort(sub_frames)
            sorted_idx = np.array(idx)[order]
            sorted_frames = sub_frames[order]

            eid = 0
            event_ids = [0]
            for k in range(1, len(sorted_frames)):
                if sorted_frames[k] - sorted_frames[k - 1] > event_gap:
                    eid += 1
                event_ids.append(eid)

            event_id_col[detections.index.get_indexer(sorted_idx)] = event_ids

        detections["event_id"] = event_id_col

        # ── SNR-weighted centroids for final groups ───────────────────────
        x_w = np.empty(len(detections))
        y_w = np.empty(len(detections))

        for sid, idx in detections.groupby("source_id").groups.items():
            if sid == -1:
                x_w[idx] = detections.loc[idx, "x"].values
                y_w[idx] = detections.loc[idx, "y"].values
                continue
            w = weights[idx]
            w_sum = w.sum()
            x_w[idx] = (detections.loc[idx, "x"].values * w).sum() / w_sum
            y_w[idx] = (detections.loc[idx, "y"].values * w).sum() / w_sum

        detections["x_w"] = x_w
        detections["y_w"] = y_w

        return detections

    def find_asteroids(self, detections: pd.DataFrame,
                       max_speed: float = 3.0,
                       min_detections: int = 3,
                       max_gap: int = 3,
                       position_tol: float = 1.5) -> tuple:
        """
        Identify linearly-moving objects (asteroids) among the detections
        using kinematic trail linking.

        The algorithm:
          1. Seeds trails from detections labelled source_id == -1 (singletons
             from group()).  If source_id is absent, all detections are used.
          2. For each seed at frame t0, finds candidate next detections in
             frames t0+1 .. t0+max_gap within max_speed * gap pixels.
          3. Extends the trail forward frame by frame: predicts the next
             position using the running velocity estimate, accepts detections
             within position_tol pixels, up to max_gap missed frames.
          4. Scores trails by number of detections and linear-fit residual.
          5. Resolves conflicts greedily (longest trail wins); detections
             claimed by a winning trail are removed from contention.

        Parameters
        ----------
        detections : DataFrame
            Must contain 'frame', 'x', 'y' columns.  Typically the output
            of group().
        max_speed : float
            Maximum asteroid speed in pixels per frame.  Default 3.0.
        min_detections : int
            Minimum number of detections to report a trail.  Default 3.
        max_gap : int
            Maximum number of consecutive missed frames when extending a
            trail.  Default 3.
        position_tol : float
            Tolerance in pixels for matching a detection to a predicted
            position on an existing trail.  Default 1.5.

        Returns
        -------
        trails : DataFrame
            One row per asteroid trail, columns:
              trail_id, n_det, t_start, t_end, t_span,
              x0, y0, vx, vy, speed,
              r2_x, r2_y          — R² of linear fit in each axis
              det_indices         — list of detection DataFrame indices
        detections : DataFrame
            Input detections with added 'trail_id' column
            (NaN = not part of any trail).
        """
        if "frame" not in detections.columns:
            raise ValueError("detections must have a 'frame' column")

        # Work only on candidate moving detections (singletons from group())
        if "source_id" in detections.columns:
            cands = detections[detections["source_id"] == -1].copy()
        else:
            cands = detections.copy()

        if len(cands) < min_detections:
            detections = detections.copy()
            detections["trail_id"] = np.nan
            return pd.DataFrame(), detections

        cands = cands.sort_values("frame").reset_index()  # keeps original idx in 'index' col
        # Build a per-frame index for fast lookup
        frames_sorted = sorted(cands["frame"].unique())
        by_frame = {t: cands[cands["frame"] == t] for t in frames_sorted}
        frame_set = set(frames_sorted)

        from scipy.spatial import KDTree

        def _kdtree(frame_t):
            sub = by_frame.get(frame_t)
            if sub is None or len(sub) == 0:
                return None, None
            return KDTree(sub[["x", "y"]].values), sub

        # ── Trail building ────────────────────────────────────────────────────
        all_trails = []  # list of (score, [original_indices])

        used_set = set()

        for _, seed in cands.iterrows():
            orig_idx = seed["index"]
            if orig_idx in used_set:
                continue
            t0 = seed["frame"]
            x0, y0 = seed["x"], seed["y"]

            # Find the best initial pair to set a starting velocity
            best_pair = None
            best_pair_dist = np.inf
            for gap in range(1, max_gap + 1):
                t1 = t0 + gap
                kdt, sub1 = _kdtree(t1)
                if kdt is None:
                    continue
                search_r = max_speed * gap
                idxs = kdt.query_ball_point([x0, y0], r=search_r)
                for j in idxs:
                    row = sub1.iloc[j]
                    dist = np.hypot(row["x"] - x0, row["y"] - y0)
                    if dist < best_pair_dist:
                        best_pair_dist = dist
                        best_pair = (t1, row["x"], row["y"], row["index"], gap)
                if best_pair is not None:
                    break  # prefer smallest gap

            if best_pair is None:
                continue  # no pair found, skip this seed

            t1, x1, y1, idx1, gap1 = best_pair
            vx = (x1 - x0) / gap1
            vy = (y1 - y0) / gap1

            # Extend the trail forward
            chain_orig = [orig_idx, idx1]
            chain_frames = [t0, t1]
            chain_x = [x0, x1]
            chain_y = [y0, y1]
            cur_vx, cur_vy = vx, vy
            cur_t = t1
            cur_x, cur_y = x1, y1
            consecutive_misses = 0

            while consecutive_misses <= max_gap:
                cur_t += 1
                if cur_t > max(frames_sorted):
                    break
                pred_x = cur_x + cur_vx
                pred_y = cur_y + cur_vy
                kdt, sub_t = _kdtree(cur_t)
                if kdt is None:
                    consecutive_misses += 1
                    cur_x += cur_vx
                    cur_y += cur_vy
                    continue
                matches = kdt.query_ball_point([pred_x, pred_y], r=position_tol)
                if not matches:
                    consecutive_misses += 1
                    cur_x += cur_vx
                    cur_y += cur_vy
                    continue
                # Pick closest match
                best_j = min(matches, key=lambda j: np.hypot(
                    sub_t.iloc[j]["x"] - pred_x,
                    sub_t.iloc[j]["y"] - pred_y))
                row = sub_t.iloc[best_j]
                chain_orig.append(row["index"])
                chain_frames.append(cur_t)
                chain_x.append(row["x"])
                chain_y.append(row["y"])
                # Update running velocity (exponential moving avg)
                dt = cur_t - chain_frames[-2]
                new_vx = (row["x"] - chain_x[-2]) / dt
                new_vy = (row["y"] - chain_y[-2]) / dt
                cur_vx = 0.7 * cur_vx + 0.3 * new_vx
                cur_vy = 0.7 * cur_vy + 0.3 * new_vy
                cur_x, cur_y = row["x"], row["y"]
                consecutive_misses = 0

            if len(chain_orig) < min_detections:
                continue

            # Linear fit quality
            frames_arr = np.array(chain_frames, dtype=float)
            x_arr = np.array(chain_x)
            y_arr = np.array(chain_y)
            r2_x = _r2(frames_arr, x_arr)
            r2_y = _r2(frames_arr, y_arr)

            score = len(chain_orig) + 0.5 * (r2_x + r2_y)
            all_trails.append((score, chain_orig, chain_frames, chain_x, chain_y,
                                r2_x, r2_y))

        # ── Conflict resolution: greedy, longest trail wins ───────────────────
        all_trails.sort(key=lambda t: -t[0])
        claimed = set()
        selected = []
        for trail in all_trails:
            score, orig_idxs, frms, xs, ys, r2x, r2y = trail
            # Skip if too many detections already claimed
            overlap = sum(1 for i in orig_idxs if i in claimed)
            if overlap > len(orig_idxs) * 0.5:
                continue
            for i in orig_idxs:
                claimed.add(i)
            selected.append(trail)

        # ── Build trail catalog ───────────────────────────────────────────────
        trail_rows = []
        trail_id_map = {}  # orig detection index → trail_id

        for tid, (score, orig_idxs, frms, xs, ys, r2x, r2y) in enumerate(selected):
            frames_arr = np.array(frms, dtype=float)
            x_arr = np.array(xs)
            y_arr = np.array(ys)
            vx_fit, x0_fit = np.polyfit(frames_arr, x_arr, 1)
            vy_fit, y0_fit = np.polyfit(frames_arr, y_arr, 1)
            t_ref = frames_arr[0]
            trail_rows.append({
                "trail_id": tid,
                "n_det": len(orig_idxs),
                "t_start": int(frms[0]),
                "t_end": int(frms[-1]),
                "t_span": int(frms[-1]) - int(frms[0]) + 1,
                "x0": float(x0_fit + vx_fit * t_ref),
                "y0": float(y0_fit + vy_fit * t_ref),
                "vx": float(vx_fit),
                "vy": float(vy_fit),
                "speed": float(np.hypot(vx_fit, vy_fit)),
                "r2_x": float(r2x),
                "r2_y": float(r2y),
                "det_indices": orig_idxs,
            })
            for i in orig_idxs:
                trail_id_map[i] = tid

        trails_df = pd.DataFrame(trail_rows) if trail_rows else pd.DataFrame(
            columns=["trail_id", "n_det", "t_start", "t_end", "t_span",
                     "x0", "y0", "vx", "vy", "speed", "r2_x", "r2_y", "det_indices"])

        detections = detections.copy()
        detections["trail_id"] = detections.index.map(trail_id_map)

        return trails_df, detections

    @property
    def background(self):
        """Background object from the last call to find(). None if not yet called."""
        return getattr(self, "background_", None)

    # ── internals ─────────────────────────────────────────────────────────────

    def _find_single(self, image: np.ndarray) -> pd.DataFrame:
        """Run multi-scale SEP detection on a single 2-D image."""
        data = np.asarray(image, dtype=np.float64)
        bkg = sep.Background(data, bw=self.bw, bh=self.bw)
        data_sub = data - bkg.back()

        if self.ref is not None:
            err = self._noise_map(bkg)
        else:
            err = bkg.rms()

        per_scale = []
        for fwhm, kern in zip(self._fwhms, self._kernels):
            try:
                objs = sep.extract(
                    data_sub,
                    thresh=self.thresh,
                    err=err,
                    minarea=2,
                    deblend_nthresh=self.deblend_nthresh,
                    deblend_cont=self.deblend_cont,
                    filter_kernel=kern,
                )
            except Exception:
                continue
            if len(objs) == 0:
                continue
            df = pd.DataFrame({
                "x": objs["x"],
                "y": objs["y"],
                "flux": objs["flux"],
                "peak": objs["peak"],
                "a": objs["a"],
                "b": objs["b"],
                "theta": objs["theta"],
                "flag": objs["flag"],
                "detection_fwhm": fwhm,
            })
            per_scale.append(df)

        if not per_scale:
            return pd.DataFrame(columns=["x", "y", "flux", "peak", "a", "b",
                                          "theta", "flag", "detection_fwhm",
                                          "snr", "ellipticity"])

        combined = (
            pd.concat(per_scale, ignore_index=True)
            .sort_values("flux", ascending=False)
            .reset_index(drop=True)
        )

        sources = self._dedup(combined)

        # ── SNR via circular aperture photometry (r = 1.5 px) ─────────────
        ap_flux, ap_err, _ = sep.sum_circle(
            data_sub, sources["x"].values, sources["y"].values,
            r=1.5, err=err, gain=1.0,
        )
        ap_err = np.where(ap_err > 0, ap_err, np.nan)
        sources["snr"] = ap_flux / ap_err

        # ── snr_model: ref-scaled noise normalisation ─────────────────────
        if self.ref is not None:
            noise_map = self._noise_map(bkg)
            xi = np.round(sources["x"].values).astype(int).clip(0, data.shape[1] - 1)
            yi = np.round(sources["y"].values).astype(int).clip(0, data.shape[0] - 1)
            sources["snr_model"] = sources["peak"].values / noise_map[yi, xi]

        # ── Ellipticity ────────────────────────────────────────────────────
        a = sources["a"].values.clip(min=1e-6)
        sources["ellipticity"] = 1.0 - sources["b"].values / a

        # ── Negative pixel extent ──────────────────────────────────────────
        sources["neg_extent"] = self._neg_pixel_extent(data_sub, sources, r=2.0)

        self.background_ = bkg
        return sources

    def _find_cube(self, cube: np.ndarray) -> pd.DataFrame:
        """Run _find_single on each frame in parallel using joblib."""
        from joblib import Parallel, delayed

        T = cube.shape[0]
        results = Parallel(n_jobs=-1, prefer="threads")(
            delayed(self._find_single)(cube[t]) for t in range(T)
        )

        frames = []
        for t, df in enumerate(results):
            if len(df) > 0:
                df = df.copy()
                df["frame"] = t
                frames.append(df)

        if not frames:
            return pd.DataFrame()
        return pd.concat(frames, ignore_index=True)

    def _noise_map(self, bkg) -> np.ndarray:
        """
        Build a per-pixel noise map combining background RMS and the
        ref-scaled systematic residual:

            noise[y,x] = sqrt(bkg_rms[y,x]² + (k * ref[y,x])²)
        """
        bkg_rms = bkg.rms()
        if self.residual_k is not None:
            k = self.residual_k
        else:
            from astropy.stats import sigma_clipped_stats
            m, med, std = sigma_clipped_stats(self.ref)
            bright = self.ref > 5 * med
            if bright.sum() > 10:
                k = float(np.median(bkg_rms[bright] / self.ref[bright]))
            else:
                k = 0.0078  # TESS default measured from sector 38
        from astropy.stats import sigma_clipped_stats
        m, med, std = sigma_clipped_stats(self.ref)
        bright = self.ref < 3*std + med
        ref_err = self.ref.copy()
        ref_err[bright] = 1
        return np.sqrt(bkg_rms**2 + 0.1*ref_err)

    @staticmethod
    def _neg_pixel_extent(data_sub: np.ndarray, sources: pd.DataFrame,
                           r: float = 1.5, box: int = 1) -> np.ndarray:
        """
        For each source, return the ratio of the most-negative pixel
        to the background stddev within a circular aperture of radius
        `r`, floored at zero.

            neg_extent = max(0, -min(pixels_in_aperture))/std
        """
        from astropy.stats import sigma_clipped_stats
        m, med, std = sigma_clipped_stats(data_sub)
        H, W = data_sub.shape
        R = int(np.ceil(r))
        dy, dx = np.mgrid[-R:R+1, -R:R+1]
        in_circle = (dx**2 + dy**2) <= r**2
        dy_offsets = dy[in_circle]
        dx_offsets = dx[in_circle]

        xs = np.round(sources["x"].values).astype(int)
        ys = np.round(sources["y"].values).astype(int)

        extents = np.empty(len(sources), dtype=float)
        for i, (cx, cy) in enumerate(zip(xs, ys)):
            px = cx + dx_offsets
            py = cy + dy_offsets
            valid = (px >= 0) & (px < W) & (py >= 0) & (py < H)
            if valid.sum() == 0:
                extents[i] = np.nan
                continue

            y0 = max(0, ys[i] - box); y1 = min(H, ys[i] + box + 1)
            x0 = max(0, xs[i] - box); x1 = min(W, xs[i] + box + 1)
            patch = data_sub[y0:y1, x0:x1]
            if patch.size == 0:
                extents[i] = np.nan
                continue
            extents[i] = max(0.0, -float(patch.min())) / std

        return extents

    @staticmethod
    def _mexhat(fwhm: float) -> np.ndarray:
        """Mexican-hat (Laplacian-of-Gaussian) kernel normalised to unit abs-sum."""
        sigma = fwhm / (2 * np.sqrt(2 * np.log(2)))
        size = max(5, int(np.ceil(6 * sigma)) | 1)
        c = size // 2
        y, x = np.mgrid[-c:c+1, -c:c+1]
        r2 = (x**2 + y**2) / (2 * sigma**2)
        k = (1 - r2) * np.exp(-r2)
        return (k / np.abs(k).sum()).astype(np.float32)

    def _dedup(self, combined: pd.DataFrame) -> pd.DataFrame:
        """Remove duplicate detections within dedup_radius, keeping highest flux."""
        from scipy.spatial import KDTree
        keep = np.ones(len(combined), dtype=bool)
        tree = KDTree(combined[["x", "y"]].values)
        for i in range(len(combined)):
            if not keep[i]:
                continue
            for j in tree.query_ball_point(combined[["x", "y"]].iloc[i].values,
                                            r=self.dedup_radius):
                if j > i:
                    keep[j] = False
        return combined[keep].reset_index(drop=True)


# ── helpers ───────────────────────────────────────────────────────────────────

def _r2(x: np.ndarray, y: np.ndarray) -> float:
    """R² of a linear fit y ~ x.  Returns 0 if fewer than 2 points."""
    if len(x) < 2 or np.ptp(x) == 0:
        return 0.0
    coeffs = np.polyfit(x, y, 1)
    y_pred = np.polyval(coeffs, x)
    ss_res = np.sum((y - y_pred) ** 2)
    ss_tot = np.sum((y - y.mean()) ** 2)
    return float(1.0 - ss_res / ss_tot) if ss_tot > 0 else 1.0
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys

print("Loading data...")
df = good#pd.read_csv('/Users/rri38/Documents/work/code/tess/test_tessellate/detected_sources.csv')
sys.path.insert(0, '/Users/rri38/Documents/work/code/tess/tessellate')
from tessellate import detector

# Clean NA
df_clean = df.copy().drop(columns=['objid', 'objid_old', 'objid_new'], errors='ignore')

# Use standard spatial group but with a generous distance to guarantee any asteroid trails are grouped together
print("Running base spatial grouping...")
df_grouped = detector._Spatial_group(df_clean, colname='obj', distance=2.0)

results = []
print("Scanning for the most linear, highly-spread actual objects...")
for oid, sub in df_grouped.groupby('obj'):
    if len(sub) > 5:
        std_x = sub['xcentroid'].std()
        std_y = sub['ycentroid'].std()
        std_t = sub['frame'].std()
        
        corr_x_t = abs(sub['xcentroid'].corr(sub['frame'])) if std_x > 0 and std_t > 0 else 0
        corr_y_t = abs(sub['ycentroid'].corr(sub['frame'])) if std_y > 0 and std_t > 0 else 0
        
        max_std = max(std_x, std_y)
        max_corr = max(corr_x_t, corr_y_t)
        
        results.append({
            'obj': oid,
            'size': len(sub),
            'max_std': max_std,
            'max_corr': max_corr,
            'corr_x': corr_x_t,
            'corr_y': corr_y_t,
            'std_x': std_x,
            'std_y': std_y
        })

df_res = pd.DataFrame(results)

# Sort by those that have both highest spatial spread AND highest correlation
# Let's define a score: max_std * max_corr
df_res['score'] = df_res['max_std'] * df_res['max_corr']
top_candidates = df_res.sort_values(by='score', ascending=False).head(4)

print(f"Top 4 moving candidates in real data:")
print(top_candidates[['obj', 'size', 'max_std', 'max_corr', 'score']])

actual_top_ids = top_candidates['obj'].tolist()

fig, axes = plt.subplots(1, len(actual_top_ids), figsize=(16, 5))
if len(actual_top_ids) == 1:
    axes = [axes]

for i, oid in enumerate(actual_top_ids):
    ax = axes[i]
    sub = df_grouped[df_grouped['obj'] == oid]
    
    # Plot background
    bg_mask = (df_grouped['xcentroid'] > sub['xcentroid'].min() - 3) & \
              (df_grouped['xcentroid'] < sub['xcentroid'].max() + 3) & \
              (df_grouped['ycentroid'] > sub['ycentroid'].min() - 3) & \
              (df_grouped['ycentroid'] < sub['ycentroid'].max() + 3) & \
              (df_grouped['obj'] != oid)
              
    ax.scatter(df_grouped[bg_mask]['xcentroid'], df_grouped[bg_mask]['ycentroid'], color='grey', s=5, alpha=0.3, label='Static Sources')
    
    # Plot candidate
    # Color by frame to show progression over time!
    sc = ax.scatter(sub['xcentroid'], sub['ycentroid'], c=sub['frame'], cmap='plasma', s=40, label=f'Obj {oid}')
    cb = plt.colorbar(sc, ax=ax)
    cb.set_label('Frame')
    
    # Check linear fit
    if len(sub) > 2:
        vx, x0 = np.polyfit(sub['frame'], sub['xcentroid'], 1)
        vy, y0 = np.polyfit(sub['frame'], sub['ycentroid'], 1)
        f_min, f_max = sub['frame'].min(), sub['frame'].max()
        x_min, x_max = vx*f_min+x0, vx*f_max+x0
        y_min, y_max = vy*f_min+y0, vy*f_max+y0
        ax.plot([x_min, x_max], [y_min, y_max], color='red', linestyle='--', label='Linear Trajectory')
        
    cand_info = top_candidates[top_candidates['obj'] == oid].iloc[0]
    ax.set_title(f"Obj {oid} | Std: {cand_info['max_std']:.2f} | Corr: {cand_info['max_corr']:.2f}")
    ax.legend(loc='lower left', prop={'size': 8})

plt.tight_layout()
#plt.savefig('/Users/rri38/Documents/work/code/tess/test_tessellate/real_asteroid_candidates.png', dpi=300)
print("Saved real_asteroid_candidates.png")
```
