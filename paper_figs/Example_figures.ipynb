```python
import tessreduce as tr
import matplotlib.pyplot as plt
%matplotlib notebook
import numpy as np

plt.rc('text', usetex=True)
plt.rc('font', family='serif')

fig_width_pt = 240.0  # Get this from LaTeX using \showthe\columnwidth
inches_per_pt = 1.0/72.27               # Convert pt to inches
golden_mean = (np.sqrt(5)-1.0)/2.0         # Aesthetic ratio
fig_width = fig_width_pt*inches_per_pt  # width in inches
```

# GRB 191016A

```python
ra = 30.26933
dec = 24.50988
sector = 17
# This function takes the sky position and downloads the relevant cutout of TESS data
grb = tr.tessreduce(ra,dec,size=90,sector=sector,reduce=True,plot=True)
```

```python
grb.lc[1] -= np.nanmedian(grb.lc[1,280:360])
```

```python
grb.to_flux()
```

```python
plt.figure(figsize=(1.5*fig_width,1*fig_width))
plt.plot(grb.lc[0]-58772,grb.lc[1],'k.')
plt.ylim(-.2,3.5)
plt.xlim(-.5,1)
plt.ylabel('Flux (mJy)',fontsize=15)
plt.xlabel(r'Time $(\rm MJD-58772)$',fontsize=15)
plt.annotate('GRB 191016A\n',(.05,.8),xycoords='axes fraction',fontsize=12)
plt.savefig('grb191016a.pdf', bbox_inches = "tight")
```

```python
grb.to_mag()
```

```python
grb.plotter()
```
